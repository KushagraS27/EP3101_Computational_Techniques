# Q1

```python
## Integration of x^3 + 1

def f(x):
    return x**3 + 1
h = []
a = 1
b = 10

h = 0.5

n = int((b - a) / h)

sum = (f(a) + f(b)) / 2

for i in range(1, n):
    x = a + i * h
    sum = sum + f(x)

trapezoidal = h * sum

# True result
true_result = (b**4 / 4 + b) - (a**4 / 4 + a)

print("Trapezoidal Result =", trapezoidal)
print("True Result =", true_result)
print("Error =", abs(true_result - trapezoidal))
```

# Plotting Curve Error Vs h values

```python

import math
import matplotlib.pyplot as plt

# Function
def f(x):
    return x**3 + 1

a = 1
b = 10

# True result
true_result = (b**4 / 4 + b) - (a**4 / 4 + a)

# h values
h_values = [3, 1.5, 1, 0.75, 0.5, 0.3, 0.25, 0.1,0.01]
#h_values=[0.01,0.1,0.2,0.3,0.4]
errors = []

for h in h_values:

    n = int((b - a) / h)

    sum = (f(a) + f(b)) / 2

    for i in range(1, n):
        x = a + i * h
        sum = sum + f(x)

    trapezoidal = h * sum

    error = (true_result - trapezoidal)

    errors.append(error)

    print("h =", h)
    print("Trapezoidal Result =", trapezoidal)
    print("True Result =", true_result)
    print("Error =", error)
    #print("------------------------")


# Graph
plt.plot(h_values, errors, marker='o')

plt.xlabel("h value")
plt.ylabel("Error")
plt.title("h Values")

plt.grid()
plt.show()

```

```python
import math

L = 4
h = 0.1

E = math.pi**2 / (2 * L**2)

x = [0, h]
psi = [0, 1]

n = int(L / h)

for i in range(1, n):
    next_psi = -psi[i-1] + (2 - 2*E*h**2)*psi[i]
    psi.append(next_psi)
    x.append((i+1)*h)

print("       x          psi")
print("----------------------")

for i in range(n+1):
    print("%8.2f     %10.6f" % (x[i], psi[i]))
plt.plot(x,psi)
plt.grid()
plt.show()
```

# 2(b)

```python
import math

L = 4
h = 0.1
n = int(L/h)

def f(x):
    return math.sin(math.pi*x/4)**2

s = f(0) + f(L)

for i in range(1, n):
    x = i*h

    if i % 2 == 0:
        s += 2*f(x)
    else:
        s += 4*f(x)

I = h*s/3

A = 1/math.sqrt(I)

print("Integral =", I)
print("Normalization constant A =", A)
```

# 2(c)

```python
import math
import matplotlib.pyplot as plt

L = 4
h = 0.1

E = math.pi**2 / (2 * L**2)

x = [0, h]
psi = [0, 1]

n = 40

for i in range(1, n):
    next_psi = -psi[i-1] + (2 - 2*E*h**2)*psi[i]
    psi.append(next_psi)
    x.append((i+1)*h)

psi_original = psi.copy()

f = []

for i in range(n+1):
    f.append(psi[i]**2)

s = f[0] + f[n]

for i in range(1, n):
    if i % 2 == 0:
        s += 2*f[i]
    else:
        s += 4*f[i]

I = h*s/3

for i in range(n+1):
    psi[i] = psi[i] / math.sqrt(I)

print("x       Original       Normalized")

for i in range(n+1):
    print("%5.2f    %10.6f    %10.6f" % (x[i], psi_original[i], psi[i]))

plt.plot(x, psi_original, label="Before Normalization")
plt.plot(x, psi, label="After Normalization")
plt.xlabel("x")
plt.ylabel("psi")
plt.title("Wave Function Before and After Normalization")
plt.legend()
plt.grid()
plt.show()
```

# Q(3)

```python
import math

b = 10000
n = 6000
h = b/n

def f(x):
    if x == 0:
        return 1
    return math.sin(x)/x

s = f(0) + f(b)

for i in range(1, n):
    x = i*h

    if i % 2 == 0:
        s += 2*f(x)
    else:
        s += 4*f(x)

I13 = h*s/3


s = f(0) + f(b)

for i in range(1, n):
    x = i*h

    if i % 3 == 0:
        s += 2*f(x)
    else:
        s += 3*f(x)

I38 = 3*h*s/8


print("Simpson 1/3 =", I13)
print("Simpson 3/8 =", I38)
print("pi/2        =", math.pi/2)

print("Error 1/3 =", abs(math.pi/2 - I13))
print("Error 3/8 =", abs(math.pi/2 - I38))
```

# 4(a)

```python
import numpy as np
import matplotlib.pyplot as plt

Q = 1
R = 1

r = np.linspace(0.01, 3, 500)

V = []

for x in r:
    if x < R:
        v = Q/(2*R)*(3 - x**2/R**2)
    else:
        v = Q/x

    V.append(v)

plt.plot(r, V)
plt.xlabel("r")
plt.ylabel("V(r)")
plt.title("Potential due to Uniformly Charged Solid Sphere")
plt.grid()
plt.show()
```
