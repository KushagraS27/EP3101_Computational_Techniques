```python
import math 
import matplotlib.pyplot as plt
## Implementing Cot(x)

# Given data
x = [1, 2, 3, 4, 5]  # degrees
f = [57.28996163, 28.63625328, 19.08113669,
     14.30066626, 11.43005230]
# x=1'
h=1
#x=float(input("Enter x Value for diffrentiation"))
## Logic Forward Difference
forward=(f[1]-f[0])/h
# x=2'
##  Logic Central Difference
central=(f[2]-f[0])/(h*2)
print("forward differnce approach :", forward)
print("centraldiffernce approach :", central)

F=[]
C=[]
## Using Loops
for i in range(4):
    forward=(f[i+1]-f[i])/h
    F.append(forward)
for j in range(1,4):
    central=(f[j+1]-f[j-1])/(2*h)
    C.append(central)
print("Forward", F)
print("Central", C)
x_forward=x[:-1]
x_central=x[1:-1]
plt.plot(x,f,label="Functio")
plt.plot(x_central,C,'^--',label="Central Difference")
plt.plot(x_forward,F,label="Forward Differnce")
plt.xlabel("X values")
plt.ylabel("Fuction value Cot(x)")
plt.title("Cot (x), Graph")
plt.legend()
plt.show()
```
