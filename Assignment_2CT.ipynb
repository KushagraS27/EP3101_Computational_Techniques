```python
# Q1 LU DECOMPOSITION FOR LINEAR EQ

n = int(input("Enter number of variables: "))


A = []

print("Enter coefficient matrix A:")

for i in range(n):
    row = list(map(float, input(f"Row {i+1}: ").split()))
    A.append(row)


print("\nEnter RHS values B:")
B = []

for i in range(n):
    B.append(float(input(f"B[{i+1}]: ")))


L = [[0.0] * n for _ in range(n)]
U = [[0.0] * n for _ in range(n)]


for i in range(n):

    
    for j in range(i, n):
        sum_value = 0

        for k in range(i):
            sum_value += L[i][k] * U[k][j]

        U[i][j] = A[i][j] - sum_value

    
    L[i][i] = 1

    for j in range(i + 1, n):
        sum_value = 0

        for k in range(i):
            sum_value += L[j][k] * U[k][i]

        L[j][i] = (A[j][i] - sum_value) / U[i][i]




Y = [0.0] * n

for i in range(n):
    sum_value = 0

    for j in range(i):
        sum_value += L[i][j] * Y[j]

    Y[i] = B[i] - sum_value




X = [0.0] * n

for i in range(n - 1, -1, -1):
    sum_value = 0

    for j in range(i + 1, n):
        sum_value += U[i][j] * X[j]

    X[i] = (Y[i] - sum_value) / U[i][i]



print("\nL Matrix:")

for row in L:
    print(["{:.4f}".format(x) for x in row])



print("\nU Matrix:")

for row in U:
    print(["{:.4f}".format(x) for x in row])



print("\nY values:")
for i in range(n):
    print(f"Y{i+1} = {Y[i]:.4f}")



print("\nSolution:")

for i in range(n):
    print(f"x{i+1} = {X[i]:.4f}")
```

```python
# Q 4 Jacobi Iteration for linear eq.




n = int(input("Enter number of variables: "))
A = []

print("Enter coefficient matrix A:")

for i in range(n):
    row = list(map(float, input(f"Row {i+1}: ").split()))
    A.append(row)


print("\nEnter RHS values:")

B = []

for i in range(n):
    B.append(float(input(f"B[{i+1}]: ")))


x = [0.0] * n


tol = float(input("\nEnter tolerance: "))
max_iter = int(input("Enter maximum number of iterations: "))


for iteration in range(max_iter):

    x_new = [0.0] * n

    for i in range(n):

        sum_value = 0.0

        for j in range(n):
            if j != i:
                sum_value += A[i][j] * x[j]

        x_new[i] = (B[i] - sum_value) / A[i][i]

    
    error = max(abs(x_new[i] - x[i]) for i in range(n))


    x = x_new

    print(f"Iteration {iteration + 1}: ", end="")

    for value in x:
        print(f"{value:.6f}", end=" ")

    print(f"  Error = {error:.6f}")

    
    if error < tol:
        print("\nConverged!")
        break

else:
    print("\nMaximum iterations reached.")


print("\nFinal Solution:")

for i in range(n):
    print(f"x{i + 1} = {x[i]:.6f}")
```

```python
# Question 3
# Schrodinger Equation for a Particle in a Box
# Using Cholesky Method

import numpy as np
import matplotlib.pyplot as plt


hbar = 1.054571817e-34
m = 9.1093837e-31       


L = float(input("Enter length of box L (m): "))
N = int(input("Order of Matrix (grid points) "))
n_states = int(input("Enter number of energy states required: "))


dx = L / (N + 1)
x = np.linspace(dx, L - dx, N)


a = hbar**2 / (2 * m * dx**2)


H = np.zeros((N, N))

for i in range(N):
    H[i, i] = 2 * a

    if i > 0:
        H[i, i - 1] = -a

    if i < N - 1:
        H[i, i + 1] = -a



Lc = np.linalg.cholesky(H)

print("\nCholesky Lower Triangular Matrix:")
print(Lc)



states = []
energies = []

for state in range(n_states):

    
    psi = np.random.rand(N)

    
    for old_psi in states:
        psi -= np.dot(old_psi, psi) * old_psi

    psi /= np.linalg.norm(psi)

    
    for iteration in range(100):

        
        y = np.linalg.solve(Lc, psi)

    
        new_psi = np.linalg.solve(Lc.T, y)

        
        for old_psi in states:
            new_psi -= np.dot(old_psi, new_psi) * old_psi

            
        new_psi /= np.linalg.norm(new_psi)

        
        if np.linalg.norm(new_psi - psi) < 1e-10:
            break

        psi = new_psi

    psi = new_psi

    
    E = np.dot(psi, H @ psi) / np.dot(psi, psi)

    
    psi = psi / np.sqrt(np.sum(psi**2) * dx)

    states.append(psi)
    energies.append(E)



print("\nEnergy Eigenvalues:")
for i, E in enumerate(energies):
    print(f"E{i+1} = {E:.6e} J")


print("\nTheoretical Energy Eigenvalues:")
for n in range(1, n_states + 1):
    E_theory = (n**2 * np.pi**2 * hbar**2) / (2 * m * L**2)
    print(f"E{n} = {E_theory:.6e} J")



plt.figure(figsize=(8, 6))

for i in range(n_states):
    plt.plot(x, states[i], label=f"n = {i+1}")

plt.xlabel("Position x (m)")
plt.ylabel(r"Wave function $\psi(x)$")
plt.title("Particle in a Box - Wave Functions")
plt.grid(True)
plt.legend()
plt.show()

"""

plt.figure(figsize=(8, 6))

for i in range(n_states):
    plt.plot(x, states[i]**2, label=f"n = {i+1}")

plt.xlabel("Position x (m)")
plt.ylabel(r"$|\psi(x)|^2$")
plt.title("Probability Density")
plt.grid(True)
plt.legend()
plt.show()
"""
```

```python
# Question 2
# Classical Harmonic Oscillator using LU Decomposition

import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Input
# ------------------------------------------------

omega = float(input("Enter angular frequency omega: "))
T = float(input("Enter total time T: "))
N = int(input("Enter number of grid points: "))

x0 = float(input("Enter displacement at t=0: "))
xT = float(input("Enter displacement at t=T: "))

# ------------------------------------------------
# Grid
# ------------------------------------------------

h = T / (N - 1)
t = np.linspace(0, T, N)

# Number of unknown interior points
n = N - 2

# ------------------------------------------------
# Construct matrix A
# ------------------------------------------------

A = np.zeros((n, n))
b = np.zeros(n)

# Coefficient of x_i
diagonal = -2 + omega**2 * h**2

for i in range(n):

    A[i, i] = diagonal

    # x_(i-1)
    if i > 0:
        A[i, i-1] = 1

    # x_(i+1)
    if i < n - 1:
        A[i, i+1] = 1

# ------------------------------------------------
# Boundary conditions
# ------------------------------------------------

# First equation contains x0
b[0] = -x0

# Last equation contains xT
b[-1] = -xT

# ------------------------------------------------
# LU Decomposition
# A = L U
# ------------------------------------------------

L = np.zeros((n, n))
U = np.zeros((n, n))

for i in range(n):

    # Calculate U
    for j in range(i, n):
        U[i, j] = A[i, j]

        for k in range(i):
            U[i, j] -= L[i, k] * U[k, j]

    # Calculate L
    L[i, i] = 1

    for j in range(i + 1, n):
        L[j, i] = A[j, i]

        for k in range(i):
            L[j, i] -= L[j, k] * U[k, i]

        L[j, i] /= U[i, i]

# ------------------------------------------------
# Forward substitution
# L y = b
# ------------------------------------------------

y = np.zeros(n)

for i in range(n):
    y[i] = b[i]

    for j in range(i):
        y[i] -= L[i, j] * y[j]

    y[i] /= L[i, i]

# ------------------------------------------------
# Backward substitution
# U x = y
# ------------------------------------------------

solution = np.zeros(n)

for i in range(n - 1, -1, -1):
    solution[i] = y[i]

    for j in range(i + 1, n):
        solution[i] -= U[i, j] * solution[j]

    solution[i] /= U[i, i]

# ------------------------------------------------
# Complete solution including boundary values
# ------------------------------------------------

X = np.zeros(N)

X[0] = x0
X[-1] = xT
X[1:-1] = solution

# ------------------------------------------------
# Display results
# ------------------------------------------------

print("\nTime and Displacement:")
for i in range(N):
    print(f"t = {t[i]:.4f}, x = {X[i]:.6f}")

print("\nStep size h =", h)

# ------------------------------------------------
# Plot
# ------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(t, X, 'o-', label="Numerical solution")

plt.xlabel("Time (t)")
plt.ylabel("Displacement x(t)")
plt.title("Classical Harmonic Oscillator using LU Decomposition")
plt.grid(True)
plt.legend()

plt.show()
```
