```python
import math
import matplotlib.pyplot as plt

X = [0.1, 0.2, 0.3, 0.4]
Y = [9.96664442, 4.93315488, 3.23272814, 2.36522242]

n = len(X)
DT = [[0 for j in range(n)] for i in range(n)]

for i in range(n):
    DT[i][0] = Y[i]

for j in range(1, n):
    for i in range(n-j):
        DT[i][j] = (X[i+j] - X[j-1]) / (DT[i+1][j-1] - DT[0][j-1])

print("Inverse Difference Table")

for i in range(n):
    for j in range(n-i):
        print(f"{DT[i][j]:12.6f}", end=" ")
    print()




x_values = []

x = 0.01

while x <= 0.4:
    x_values.append(x)
    x = x + 0.001




y = []
e_rational = []

for x in x_values:

    p = DT[0][n-1]

    for i in range(n-2, -1, -1):
        p = DT[0][i] + (x-X[i])/p

    actual = math.cos(x)/math.sin(x)

    error = actual-p

    y.append(p)
    e_rational.append(error)




A = []
e_lagrange = []

for x in x_values:

    p = 0

    for i in range(4):

        l = 1

        for j in range(4):

            if i != j:
                l = l*(x-X[j])/(X[i]-X[j])

        p = p + l*Y[i]

    actual = math.cos(x)/math.sin(x)

    error = actual-p

    A.append(p)
    e_lagrange.append(error)

print("X Values:", x_values)
print("Rational Interpolated Values:", y)
print("Lagrange Interpolated Values:", A)
plt.plot(x_values, y, label="Rational Interpolation")
plt.plot(x_values, A, label="Lagrange Interpolation")
plt.plot(X, Y, 'o', label="Data Points")

plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Rational vs Lagrange Interpolation")
plt.legend()
plt.show()
```

```python
import math
import matplotlib.pyplot as plt

X = [0, math.pi/2, math.pi]
Y = [0, 1, 0]

X_values = []
actual_error = []
error_bound = []

x = 0

while x <= math.pi:

# Lagrange interpolation
    P = 0

    for i in range(3):
        L = 1

        for j in range(3):
            if i != j:
                L = L * (x - X[j]) / (X[i] - X[j])

        P = P + Y[i] * L

    
    E = abs(math.sin(x) - P)

# Theoretical error bound
    
    bound = abs(x * (x - math.pi/2) * (x - math.pi)) / math.factorial(3)

    X_values.append(x)
    actual_error.append(E)
    error_bound.append(bound)

    x = x + 0.01



plt.plot(X_values, actual_error, label="Actual Error")
plt.plot(X_values, error_bound, '--', label="Theoretical Error Bound")

plt.xlabel("x")
plt.ylabel("Error")
plt.title("Actual Error vs Theoretical Error Bound")
plt.legend()
plt.grid()
plt.show()
```
